# 01 — SEC Data Collection

Pull raw Company Facts data from the SEC for:

- Walmart
- Target
- Costco
- Home Depot
- Lowe's


```python
import pandas as pd
import requests
import json
import os
import time
from pathlib import Path
```

```python
PROJECT_ROOT = Path("..")
RAW_DIR = PROJECT_ROOT / "data" / "raw"
RAW_DIR.mkdir(parents=True, exist_ok=True)

companies = {
    "Walmart": "0000104169",
    "Target": "0000027419",
    "Costco": "0000909832",
    "Home Depot": "0000354950",
    "Lowes": "0000060667"
}

companies
```

## SEC request header

Replace the placeholder email before running SEC requests.


```python
headers = {
    "User-Agent": "Nisha Rajkumar nisha.rajkumar.offl@gmail.com"
}
```

```python
def download_company_facts(company, cik):
    url = f"https://data.sec.gov/api/xbrl/companyfacts/CIK{cik}.json"

    response = requests.get(url, headers=headers, timeout=30)
    response.raise_for_status()

    data = response.json()

    filename = company.lower().replace(" ", "_").replace("'", "") + ".json"
    file_path = RAW_DIR / filename

    with open(file_path, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2)

    return file_path
```

```python
downloaded_files = []

for company, cik in companies.items():
    try:
        path = download_company_facts(company, cik)
        downloaded_files.append(path)
        print(f"{company}: downloaded -> {path}")
    except Exception as e:
        print(f"{company}: failed -> {e}")

    time.sleep(0.2)
```
